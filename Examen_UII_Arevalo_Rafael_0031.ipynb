```python
# Clase Computadora
class Computadora:
    # Constructor
    def __init__(self, procesador, memoriaRAM, almacenamiento):
        self.procesador = procesador
        self.memoriaRAM = memoriaRAM
        self.almacenamiento = almacenamiento

    # Función para leer datos
    def leerDatos(self):
        self.procesador = input("Procesador: ")
        self.memoriaRAM = input("Memoria RAM: ")
        self.almacenamiento = input("Almacenamiento: ")

    # Función para mostrar datos
    def mostrarDatos(self):
        print("Procesador:", self.procesador)
        print("Memoria RAM:", self.memoriaRAM)
        print("Almacenamiento:", self.almacenamiento)


# -------- COMPUTADORA 1 --------
print("COMPUTADORA 1")
pc1 = Computadora("", "", "")
pc1.leerDatos()

# -------- COMPUTADORA 2 --------
print("\nCOMPUTADORA 2")
pc2 = Computadora("", "", "")
pc2.leerDatos()

# -------- MOSTRAR DATOS --------
print("\nDATOS COMPUTADORA 1")
pc1.mostrarDatos()

print("\nDATOS COMPUTADORA 2")
pc2.mostrarDatos()

print("\nPrograma terminado y generado por Arevalo Rafael 0031")
```
